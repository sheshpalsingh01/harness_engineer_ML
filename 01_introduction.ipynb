### What is Harness?

**Harness** is a modern, AI-native **Software Delivery Platform** that automates your entire CI/CD pipeline.

Think of it as the **autopilot for shipping code** — from your laptop to production servers.

```
You Write Code → Harness Builds → Harness Tests → Harness Deploys → Users See It
     (CI)            (CI)            (CI)              (CD)          (Done! 🎉)
```

## 🔑 Core Concepts (Must Know!)

| Term | Definition | Real-World Analogy |
|------|-----------|--------------------|
| **Pipeline** | Complete workflow (build → test → deploy) | Assembly Line |
| **Stage** | Major phase within a pipeline | Station on the assembly line |
| **Step** | Individual task within a stage | One worker's task |
| **Connector** | Connection to external tools (GitHub, Docker Hub) | Power Plug/Adapter |
| **Delegate** | Agent installed in YOUR infrastructure | Ground Crew at Airport |
| **Service** | The app you're deploying | The product you're shipping |
| **Environment** | Where you deploy (dev, staging, prod) | Factory locations |
| **Artifact** | Built package (Docker image, wheel) | Packaged product |
| **Trigger** | What starts the pipeline (git push, cron) | Factory alarm bell |
| **Template** | Reusable pipeline blueprint | Standard Operating Procedure |
| **Secret** | Encrypted credentials | Bank vault |

## 🧱 Pipeline Building Blocks

```
Pipeline (The whole workflow)
├── Stage 1: Build (CI)
│   ├── Step 1: Clone Repository
│   ├── Step 2: pip install -r requirements.txt
│   ├── Step 3: pytest (run tests)
│   └── Step 4: docker build (create image)
├── Stage 2: Deploy to Staging (CD)
│   ├── Step 1: kubectl apply (deploy to K8s)
│   └── Step 2: Smoke Tests
└── Stage 3: Deploy to Production (CD)
    ├── Step 1: Canary Deploy (10% traffic)
    ├── Step 2: Verify Health
    └── Step 3: Full Rollout (100% traffic)
```

````python
# Let's visualize the Harness hierarchy with Python

harness_hierarchy = {
    "Organization": "MyCompany",
    "Projects": [
        {
            "name": "ML-Platform",
            "pipelines": [
                {
                    "name": "Train & Deploy Model",
                    "stages": [
                        {"name": "Build", "type": "CI", "steps": ["pip install", "pytest", "docker build"]},
                        {"name": "Deploy Staging", "type": "CD", "steps": ["k8s deploy", "smoke test"]},
                        {"name": "Deploy Prod", "type": "CD", "steps": ["canary 20%", "verify", "rollout 100%"]},
                    ]
                }
            ]
        }
    ]
}

# Pretty print the structure
def print_hierarchy(data, indent=0):
    prefix = "  " * indent
    if isinstance(data, dict):
        for key, value in data.items():
            if isinstance(value, (dict, list)):
                print(f"{prefix}📁 {key}:")
                print_hierarchy(value, indent + 1)
            else:
                print(f"{prefix}🔹 {key}: {value}")
    elif isinstance(data, list):
        for item in data:
            if isinstance(item, dict) and 'name' in item:
                print(f"{prefix}├── {item['name']}")
                print_hierarchy({k: v for k, v in item.items() if k != 'name'}, indent + 1)
            elif isinstance(item, str):
                print(f"{prefix}  • {item}")
            else:
                print_hierarchy(item, indent + 1)

print("🏢 HARNESS HIERARCHY")
print("=" * 50)
print_hierarchy(harness_hierarchy)
````

## 📝 YAML Pipeline Syntax — The Language of Harness

Every Harness pipeline is written in YAML. Here's the basic structure:

```yaml
pipeline:
  name: My ML Pipeline              # Human-readable name
  identifier: my_ml_pipeline         # Unique ID (no spaces)
  projectIdentifier: ml_project      # Which project
  orgIdentifier: default             # Which organization
  
  stages:
    - stage:
        name: Build and Test
        type: CI                     # CI = build/test
        spec:
          execution:
            steps:
              - step:
                  type: Run
                  name: Run Tests
                  spec:
                    shell: Sh
                    command: pytest --verbose
    
    - stage:
        name: Deploy
        type: Deployment             # CD = deploy
        spec:
          deploymentType: Kubernetes
          execution:
            steps:
              - step:
                  type: K8sRollingDeploy
                  name: Rolling Deploy
```

````python
# Let's understand YAML pipelines with Python
import json

# A Harness pipeline represented as Python dict
# (YAML and Python dicts map 1:1)

pipeline_config = {
    "pipeline": {
        "name": "ML Model Pipeline",
        "identifier": "ml_model_pipeline",
        "stages": [
            {
                "stage": {
                    "name": "Build & Test",
                    "type": "CI",
                    "steps": [
                        {"type": "Run", "command": "pip install -r requirements.txt"},
                        {"type": "Run", "command": "pytest tests/ --verbose"},
                        {"type": "BuildAndPushDockerRegistry", "repo": "myuser/ml-model"},
                    ]
                }
            },
            {
                "stage": {
                    "name": "Deploy to K8s",
                    "type": "Deployment",
                    "deploymentType": "Kubernetes",
                    "steps": [
                        {"type": "K8sRollingDeploy", "skipDryRun": False},
                    ]
                }
            }
        ]
    }
}

print("📋 Pipeline Configuration (Python dict → YAML)")
print("=" * 50)
print(json.dumps(pipeline_config, indent=2))
print("\n💡 TIP: This dict maps directly to Harness YAML!")
````

## 🚀 Deployment Strategies Explained

### 1. Rolling Deployment (Default)
```
[Pod1 v1] [Pod2 v1] [Pod3 v1]   ← Start
[Pod1 v2] [Pod2 v1] [Pod3 v1]   ← Replace 1
[Pod1 v2] [Pod2 v2] [Pod3 v1]   ← Replace 2
[Pod1 v2] [Pod2 v2] [Pod3 v2]   ← Done ✅
```

### 2. Canary Deployment (Safest)
```
[Pod1 v1] [Pod2 v1] [Pod3 v1] [Canary v2]  ← 10% traffic to v2
... verify health ...
[Pod1 v2] [Pod2 v2] [Pod3 v2]               ← Full rollout ✅
```

### 3. Blue-Green (Zero Downtime)
```
🔵 BLUE (Live)  [v1] [v1]  ← Serving traffic
🟢 GREEN (Idle)  [v2] [v2]  ← Ready, no traffic
... swap traffic ...
🟢 GREEN (Live)  [v2] [v2]  ← Now serving ✅
```

````python
# Simulate deployment strategies in Python
import time

def simulate_rolling_deploy(pods=4, new_version="v2"):
    """Simulate a rolling deployment"""
    state = ["v1"] * pods
    print("🔄 ROLLING DEPLOYMENT")
    print(f"   Start:  {state}")
    
    for i in range(pods):
        state[i] = new_version
        print(f"   Step {i+1}: {state}  ← Pod {i+1} updated")
    
    print(f"   ✅ Done! All pods running {new_version}\n")

def simulate_canary_deploy(pods=4, canary_pct=25):
    """Simulate a canary deployment"""
    canary_count = max(1, pods * canary_pct // 100)
    print("🐤 CANARY DEPLOYMENT")
    print(f"   Start:   {['v1'] * pods}")
    
    state = ["v1"] * pods + ["v2"] * canary_count
    print(f"   Canary:  {state}  ← {canary_pct}% traffic to v2")
    print(f"   Verify:  Checking health... ✅ Healthy!")
    print(f"   Rollout: {['v2'] * pods}  ← 100% rollout")
    print(f"   ✅ Done!\n")

def simulate_bluegreen_deploy(pods=3):
    """Simulate a blue-green deployment"""
    print("🔵🟢 BLUE-GREEN DEPLOYMENT")
    print(f"   🔵 Blue (LIVE):  {['v1'] * pods}  ← Serving traffic")
    print(f"   🟢 Green (IDLE): {['v2'] * pods}  ← Deployed, no traffic")
    print(f"   🔄 Swap traffic...")
    print(f"   🔵 Blue (IDLE):  {['v1'] * pods}  ← Standby")
    print(f"   🟢 Green (LIVE): {['v2'] * pods}  ← Now serving! ✅\n")

simulate_rolling_deploy()
simulate_canary_deploy()
simulate_bluegreen_deploy()
````
